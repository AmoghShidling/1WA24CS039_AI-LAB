```python

def get_state(message):
    while True:
        state = tuple(map(int, input(message).split()))
        if len(state) == 9 and set(state) == set(range(9)):
            return state
        print("Enter numbers 0 to 8 exactly once.")


def successors(state):
    zero = state.index(0)
    r, c = divmod(zero, 3)
    result = []

    for nr, nc in [(r-1,c), (r+1,c), (r,c-1), (r,c+1)]:
        if 0 <= nr < 3 and 0 <= nc < 3:
            j = nr * 3 + nc
            s = list(state)
            s[zero], s[j] = s[j], s[zero]
            result.append(tuple(s))

    return result


def display(path):
    for step, state in enumerate(path):
        print("Step", step)
        for i in range(0, 9, 3):
            print(state[i:i+3])
        print()
    print("Total moves:", len(path) - 1)


def dfs(start, goal):
    stack = [(start, [start])]
    visited = set()

    while stack:
        state, path = stack.pop()

        if state == goal:
            print("\nDFS Solution")
            display(path)
            return

        if state in visited:
            continue

        visited.add(state)

        for nxt in successors(state):
            if nxt not in visited:
                stack.append((nxt, path + [nxt]))

    print("Solution not found")


start = get_state("Enter initial state: ")
goal = get_state("Enter goal state: ")
dfs(start, goal)
```

```python

def get_state(message):
    while True:
        state = tuple(map(int, input(message).split()))
        if len(state) == 9 and set(state) == set(range(9)):
            return state
        print("Enter numbers 0 to 8 exactly once.")


def successors(state):
    zero = state.index(0)
    r, c = divmod(zero, 3)

    for nr, nc in [(r-1,c), (r+1,c), (r,c-1), (r,c+1)]:
        if 0 <= nr < 3 and 0 <= nc < 3:
            j = nr * 3 + nc
            s = list(state)
            s[zero], s[j] = s[j], s[zero]
            yield tuple(s)


def display(path):
    for step, state in enumerate(path):
        print("Step", step)
        for i in range(0, 9, 3):
            print(state[i:i+3])
        print()
    print("Total moves:", len(path) - 1)


def dls(state, goal, limit, path):
    if state == goal:
        return path

    if limit == 0:
        return None

    for nxt in successors(state):
        if nxt not in path:
            result = dls(nxt, goal, limit - 1, path + [nxt])
            if result is not None:
                return result

    return None


def ids(start, goal):
    depth = 0

    while True:
        print("Searching depth:", depth)
        result = dls(start, goal, depth, [start])

        if result is not None:
            print("\nIDS Solution")
            display(result)
            return

        depth += 1


start = get_state("Enter initial state: ")
goal = get_state("Enter goal state: ")
ids(start, goal)
```
