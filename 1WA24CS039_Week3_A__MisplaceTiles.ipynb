```python

import heapq


def get_state(message):
    while True:
        state = tuple(map(int, input(message).split()))
        if len(state) == 9 and set(state) == set(range(9)):
            return state
        print("Enter numbers 0 to 8 exactly once.")


def successors(state):
    zero = state.index(0)
    r, c = divmod(zero, 3)

    for nr, nc in [(r-1,c), (r+1,c), (r,c-1), (r,c+1)]:
        if 0 <= nr < 3 and 0 <= nc < 3:
            j = nr * 3 + nc
            s = list(state)
            s[zero], s[j] = s[j], s[zero]
            yield tuple(s)


def misplaced(state, goal):
    return sum(1 for i in range(9)
               if state[i] != 0 and state[i] != goal[i])


def display(path):
    for step, state in enumerate(path):
        print("Step", step)
        for i in range(0, 9, 3):
            print(state[i:i+3])
        print()
    print("Total moves:", len(path) - 1)


def astar(start, goal):
    pq = [(misplaced(start, goal), 0, start, [start])]
    best = {start: 0}

    while pq:
        f, g, state, path = heapq.heappop(pq)

        if g != best[state]:
            continue

        if state == goal:
            print("\nA* Solution - Misplaced Tiles")
            display(path)
            return

        for nxt in successors(state):
            ng = g + 1

            if ng < best.get(nxt, float("inf")):
                best[nxt] = ng
                nf = ng + misplaced(nxt, goal)
                heapq.heappush(pq, (nf, ng, nxt, path + [nxt]))

    print("Solution not found")


start = get_state("Enter initial state: ")
goal = get_state("Enter goal state: ")
astar(start, goal)
```
