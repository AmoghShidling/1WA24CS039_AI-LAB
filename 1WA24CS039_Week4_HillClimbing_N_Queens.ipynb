```python

from itertools import combinations

# Calculate attacking pairs
def cost(state):
    count = 0

    for i, j in combinations(range(4), 2):
        if abs(i - j) == abs(state[i] - state[j]):
            count += 1

    return count


# Generate neighbors by swapping two queens
def neighbors(state):
    result = []

    for i, j in combinations(range(4), 2):
        new_state = list(state)
        new_state[i], new_state[j] = new_state[j], new_state[i]
        result.append(((i, j), tuple(new_state)))

    return result


# Display chessboard
def display(state):
    for row in range(4):
        for col in range(4):
            print("Q" if state[col] == row else ".", end=" ")
        print()


# Hill Climbing algorithm
def hill_climbing(current):
    while True:
        current_cost = cost(current)

        print("\nCurrent state:", current)
        print("Current cost:", current_cost)

        display(current)

        if current_cost == 0:
            print("Goal reached!")
            return current

        all_neighbors = neighbors(current)

        for pair, state in all_neighbors:
            print("Swap", pair, ":", state, "Cost =", cost(state))

        pair, best = min(
            all_neighbors,
            key=lambda item: (cost(item[1]), item[0])
        )

        if cost(best) >= current_cost:
            print("No better neighbor. Stopping.")
            return current

        print("Selected swap:", pair)
        current = best


# Main program: User input
print("HILL CLIMBING - 4 QUEENS")
print("Enter 4 row positions (0 to 3), one for each column.")
print("Example: 3 1 2 0")

state = tuple(map(int, input("Enter initial state: ").split()))

if len(state) != 4 or any(row < 0 or row > 3 for row in state):
    print("Invalid input! Enter exactly 4 numbers between 0 and 3.")
elif len(set(state)) != 4:
    print("Invalid input! Each row must be unique for this implementation.")
else:
    solution = hill_climbing(state)

    print("\nFinal state:", solution)
    print("Final cost:", cost(solution))

    if cost(solution) == 0:
        print("Solution found!")
    else:
        print("Local minimum reached; no better neighbor found.")
```
