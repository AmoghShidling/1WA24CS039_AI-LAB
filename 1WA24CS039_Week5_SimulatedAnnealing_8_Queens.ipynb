```python

import random
import math

# Calculate number of attacking queen pairs
def cost(state):
    count = 0
    for i in range(8):
        for j in range(i + 1, 8):
            if state[i] == state[j] or abs(state[i] - state[j]) == abs(i - j):
                count += 1
    return count

# Display the chessboard
def display(state):
    for row in range(8):
        for col in range(8):
            print("Q" if state[col] == row else ".", end=" ")
        print()

# Simulated Annealing
def simulated_annealing(T, cooling):
    current = [random.randint(0, 7) for _ in range(8)]

    print("\nInitial State:", current)
    print("Initial Cost:", cost(current))

    while T > 0.01 and cost(current) != 0:

        # Generate random neighbor
        next_state = current.copy()
        col = random.randint(0, 7)
        next_state[col] = random.choice(
            [r for r in range(8) if r != current[col]]
        )

        # Calculate energy difference
        delta = cost(current) - cost(next_state)

        # Accept better or probabilistically worse state
        if delta > 0 or random.random() < math.exp(delta / T):
            current = next_state

        # Decrease temperature
        T = T * cooling

    return current


# Main Program
T = float(input("Enter initial temperature (100): "))
cooling = float(input("Enter cooling rate (0.99): "))

if T <= 0 or not 0 < cooling < 1:
    print("Invalid temperature or cooling rate.")
else:
    solution = simulated_annealing(T, cooling)

    print("\nFinal State:", solution)
    print("Final Cost:", cost(solution))
    print("\nFinal Chessboard:")
    display(solution)

    if cost(solution) == 0:
        print("\nGoal Reached!")
    else:
        print("\nGoal not reached. Try running again.")
```
